```python
import mysql.connector

dataBase = mysql.connector.connect(
  host="localhost",
  user="root",
  passwd="",
  database="testds"
)

print("Database Created")
```

```python
# Create a Cursor
cursor = dataBase.cursor()

#Table Creation

student = """
CREATE TABLE STUDENT(
NAME VARCHAR(20) NOT NULL,
BRANCH VARCHAR(50),
ROLL INT NOT NULL,
SECTION VARCHAR(5),
AGE INT
)
"""

cursor.execute(student)
print("Table Created")
```

```python
query = """
INSERT INTO STUDENT (NAME, BRANCH, ROLL, SECTION, AGE)
VALUES ('KARAN','MCA',101,'A',22)
"""

cursor.execute(query)
dataBase.commit()
print("student inserted successfully")
```

```python
cursor.execute("""
select * from student
""")

rows = cursor.fetchall()
for row in rows:
  print(row)
```

```python
query = """
INSERT INTO STUDENT(NAME, BRANCH, ROLL, SECTION,AGE)
VALUES(%s,%s,%s,%s,%s)

"""

students = [
  ('Anas','MCA',102,'B',21),
  ('Aman','MCA',103,'A',23),
  ('Saad','MCA',104,'C',22)
]
cursor.executemany(query,students)
dataBase.commit()
print("Multiple Values inserted")
```
