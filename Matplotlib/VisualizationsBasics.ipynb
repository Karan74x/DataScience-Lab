```python
import matplotlib.pyplot as plt
x = [0,1,2,3,4]
y = [0,1,4,9,16]
plt.plot(x,y)  #by default Line chart
plt.show()
```

```python
x = [10, 20, 30, 40]
y = [20, 25, 35, 55]
plt.plot(x, y)
plt.title("Line Chart")
plt.ylabel('Y-Axis')
plt.xlabel('X-Axis')
plt.show()
```

```python
x = [7, 8, 9, 10, 10, 12, 12, 12, 13, 14, 14, 15, 16, 16, 17, 18, 18, 19, 20, 20,
     21, 22, 23, 24, 25, 25, 26, 28, 30, 32, 35, 36, 38, 40, 42, 44, 48, 50]

plt.hist(x, bins=10, rwidth=0.8, color='steelblue')  #hsitogram
plt.title("Histogram")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.plot()
```

```python
x = [10, 15, 20, 25, 30]
y = [12, 18, 25, 28, 35]

plt.scatter(x, y) #Scatter Plots
plt.title("Scatter Plot")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
cars = ['AUDI', 'BMW', 'FORD','TESLA', 'JAGUAR',]
data = [23, 10, 35, 15, 12]
plt.pie(data, labels=cars, autopct='%1.1f%%') #Pie-Chart
plt.title(" Pie Chart")
plt.show()
```
