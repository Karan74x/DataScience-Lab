```python
# 4. Create a working copy named df_clean before making conversions.
df_clean = df.copy()
df_clean
```

```python
# Part B — Clean Text Before Conversion
# 1. Convert Name to Pandas string dtype and remove leading/trailing spaces.
df_clean["Name"] = df_clean["Name"].astype("string")
df_clean["Name"] = df_clean["Name"].str.strip().str.title()
print(df_clean["Name"])

# 2. Inspect the unique values in Hostel before converting it.
df_clean["Hostel"]

```

```python
# 1. Convert Age to numeric safely. Invalid ages must become missing values rather than causing the program to stop.
df_clean["Age"] = pd.to_numeric(df_clean["Age"], errors='coerce')

# 2. Convert Marks to numeric safely. Identify which original values became NaN.
df_clean["Marks"] = pd.to_numeric(df_clean["Marks"], errors='coerce')
df_clean
# 3. Convert Semester to a nullable integer dtype.
df_clean["Semester"] = df_clean["Semester"].astype("int64")
df_clean["Semester"]
# 4. Convert Age to nullable Int64 after numeric conversion.
df_clean["Age"] = df_clean["Age"].fillna(df_clean["Age"].median()).astype("int64")
df_clean["Age"]

# 5. Display Name, Age, Marks, and Semester after conversion
df_clean[["Name", "Age", "Marks", "Semester"]]
```
