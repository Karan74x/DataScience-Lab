```python
import pandas as pd

df = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112],
    "Name": ["Amit", "Neha", "Rahul", "Priya", "Karan", "Diya",
             "Rohan", "Meera", "Arjun", "Kavya", "Jay", "Isha"],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "IT",
                   "Sales", "IT", "Finance", "Sales", "Finance", "HR"],
    "Experience": [2, 5, 3, 7, 4, 1, 6, 8, 5, 2, 9, 3],
    "Salary": [35000, 45000, 40000, 60000, 42000, 30000,
               55000, 75000, 65000, 38000, 85000, 37000],
    "Performance": ["Good", "Average", "Excellent", "Good", "Average",
                    "Good", "Excellent", "Excellent", "Good", "Average",
                    "Excellent", "Good"],
    "Joining_Date": ["2022-01-15", "2020-06-10", "2021-03-20",
                     "2019-08-12", "2021-11-05", "2023-02-18",
                     "2020-09-25", "2018-04-16", "2019-12-01",
                     "2022-07-11", "2017-05-08", "2021-06-22"]
})

df
```

```python
# initial inspection
print(df.head()) #first 5 rows
print(df.shape) # shows rows, columns
print(df.dtypes) #shows datatype of each columns
# print(df.info()) #shows columns and their data types
print(df.isnull().sum()) #count the number of null values fro  each column
df.describe() #Gives statistical information about numerical columns
```

```python
# Find the average salary of all employees.
print(df["Salary"].mean())


```

```python
# Find the highest-paid employee.
print(df["Salary"].max())

```

```python
# Count employees in each department.
print(df.groupby("Department")["Employee_ID"].count())


```

```python
# Calculate the average salary department-wise.
print(df.groupby("Department")["Salary"].mean())

```

```python
# Find employees with more than 4 years of experience.
print(df[df["Experience"] > 4])

```

```python
# Sort employees by salary from highest to lowest.
print(df.sort_values(by="Salary" ,ascending=False))
```

```python
# Find the department with the highest average salary.
df.groupby("Department")["Salary"].mean().max()
```

```python
# Calculate the total salary expenditure for each department.
print(df.groupby("Department")["Salary"].sum())
```

```python
# Convert Joining_Date to datetime and find employees who joined after January 1, 2021.

df["Joining_Date"] = pd.to_datetime(df["Joining_Date"], errors='coerce', dayfirst=True, format='mixed')

df[ df["Joining_Date"] > '2021-01-01']
```
