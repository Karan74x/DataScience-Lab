```python
import numpy as np

#create an ndarray from list
data =[1,2,3,4,5]
array = np.array(data)

print([1,2,3,4,5])
# no commas between numbers, hat's how you visually tell a NumPy array apart from a Python list
print(array)
print(type(array)) #type-> numpy.ndarray
```

```python
# Numpy array must have same datatype.
# If we mix int with string/float, everythng gets converted to string/float.
ar = np.array([1,2,3,4,'99.9'])


ar
```

```python
# dtype forcec a specific data type
arr = np.array([1,2,3], dtype=float)
arr
```

```python
# Topic 2: Accessing Elements (Indexing)

# arr[index]              # 1-D array
# arr[row, col]           # 2-D array
# arr[dim1, dim2, dim3]   # 3-D array
# arr[-1]                 # negative indexing → from the end
```

```python
#        1D Array
ar = np.array([10,20,40,74])
print(ar[0]) #First Element
print(ar[3]) #Last Element
```

```python
#       2D Array

ar2d = np.array([[1,2,3,4], [5,6,7,8]])
print(ar2d[0, 1]) # 0th row , 1st element, also [-2, -3]
print(ar2d[1, 2]) # 1st row , 2nd element, also grid[-1, -3]
print(ar2d[0, -2]) #0th row, last 2nd element
```
