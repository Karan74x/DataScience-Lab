```python
import pandas as pd
df = pd.read_csv("case1_student_admission_raw.csv")
```

```python
# Part A
# Q1.Load the raw CSV into df and display the first 10 rows.
print(df.head(10))
```

```python
# 2. Check df.dtypes and df.info(). Which columns are not stored in the most useful type?
print(df.dtypes)
print(df.info())
```

```python
# 3. Identify the values in Age, Marks, Admission_Date, Hostel, and Semester that may cause conversion problems.
print(df["Age"].unique())
print(df["Marks"].unique())
print(df["Admission_Date"].unique())
print(df["Hostel"].unique())
print(df["Semester"].unique())
```

```python
# 4. Create a working copy named df_clean before making conversions.
df_clean = df.copy()
df_clean
```

```python
# Part B — Clean Text Before Conversion
# 1. Convert Name to Pandas string dtype and remove leading/trailing spaces.
df_clean["Name"] = df_clean["Name"].astype("string")
df_clean["Name"] = df_clean["Name"].str.strip()
for name in df_clean["Name"]:
  print(repr(name))
```

```python
# 2. Inspect the unique values in Hostel before converting it.
df_clean["Hostel"].unique()

# 3. Explain why strip() is useful before numeric or Boolean conversion.
# because it remove the unwanted spaces so that , no spaces are left in values
```

```python
#                               Part C — Numeric Conversion
# 1. Convert Age to numeric safely. Invalid ages must become missing values rather than causing the program to stop.
df_clean["Age"] = pd.to_numeric(df_clean["Age"], errors='coerce')

# 2. Convert Marks to numeric safely.
df_clean["Marks"] = pd.to_numeric(df_clean["Marks"], errors='coerce')
# &  Identify which original values became NaN.
df_clean["Marks"].isna()


# 3. Convert Semester to a nullable integer dtype.
df_clean["Semester"] = df_clean["Semester"].astype("Int64")
df_clean["Semester"]

# 4. Convert Age to nullable Int64 after numeric conversion.
df_clean["Age"] = df_clean["Age"].astype("Int64")
df_clean["Age"]

# 5. Display Name, Age, Marks, and Semester after conversion.
df_clean[["Name", "Age", "Marks", "Semester"]]
```

```python
# Part D — Date and Boolean Conversion
# 1. Convert Admission_Date to datetime. The file contains mixed day-first formats and one invalid date.
df_clean["Admission_Date"] = pd.to_datetime(df_clean["Admission_Date"], errors='coerce', format='mixed', dayfirst=True)
df_clean["Admission_Date"]

# 2. Find the row whose Admission_Date became NaT.
df_clean[df_clean["Admission_Date"].isna()]

# 3. Standardize Hostel values and convert them to Pandas nullable boolean dtype.
df_clean["Hostel"] = df_clean["Hostel"].str.strip()
df_clean["Hostel"] = df_clean["Hostel"].replace({
  "YES": "Yes", "yes": "Yes", "Y":"Yes", "N":"No", "TRUE":"Yes", "FALSE":"No"
})

df_clean["Hostel"] = df_clean["Hostel"].map({'Yes':True, "No": False}).astype("boolean")
df_clean["Hostel"]

# 4. Display Name, Admission_Date, and Hostel.
df_clean[["Name", "Admission_Date", "Hostel"]]
```

```python
# Part E — Category Conversion
# 1. Convert Course to category dtype.
df_clean["Course"] = df_clean["Course"].astype("category")
df_clean["Course"]
# 2. Convert City to category dtype.
df_clean["City"] = df_clean["City"].astype("category")
df_clean["City"]
# 3. Display the categories available in Course and City.
print(df_clean["Course"])
print(df_clean["City"])
# 4. Explain why Course and City are better candidates for category than Student_ID.
# BEcause Student iD is mostly fall in the category of numbers but the city and Course are proper categoris and we can perform grouping on them but not on student ID
```

```python
# Part F — Verification and Analysis
# 1. Display the final dtypes.
df_clean.dtypes
# 2. Count missing values created by failed conversions.
print(df_clean.isna().sum())

# 3. Display all rows containing a missing Age, Marks, orAdmission_Date after conversion. ( use ' | ' ) 
print(
    df_clean[
        df_clean["Age"].isna()
        | df_clean["Marks"].isna()
        | df_clean["Admission_Date"].isna()
    ]
)
# 4. Compare the raw and converted Marks columns and explain what errors='coerce' did.
print(df["Marks"])
print(df_clean["Marks"])

# Errors='coerce'->  converts invalid values (like trying to convert "abc" into a number ) into missing data
# 5. Save your converted DataFrame as your own CSV and compare it with the supplied clean dataset only after completing the practical.
df_clean.to_csv("case1_clean.csv", index=False)

```
