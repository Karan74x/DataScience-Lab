```python
import pandas as pd

df = pd.read_csv("employee_scaling_practice.csv")

#            Z-SCORE

# 1st take the mean value of the column
salary_mean = df["Salary"].mean()
salary_mean

# 2nd , take the standart deviation of the column
# Standard deviation tells us how spread out the salaries are from the average.
salary_std = df["Salary"].std()
salary_std

df["Salary_ZScore"] = round(((df["Salary"]-salary_mean)/salary_std),2)
df
```
